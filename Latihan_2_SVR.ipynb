# Latihan Scikit-learn: Support Vector Regression (SVR)

Latihan ini menggunakan dataset pengalaman kerja dan gaji (`Salary_Data.csv`). Model mempelajari hubungan antara `YearsExperience` sebagai fitur dan `Salary` sebagai target numerik.

**Persiapan:** upload file `Salary_Data.csv` ke Google Colab sebelum menjalankan cell pembacaan data.

```python
# Import library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.svm import SVR
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## 1. Membaca dataset

Baca file CSV dan pastikan kolom `YearsExperience` serta `Salary` tersedia.

```python
# Pastikan Salary_Data.csv sudah di-upload ke Colab
data = pd.read_csv('Salary_Data.csv')

required_columns = {'YearsExperience', 'Salary'}
missing_columns = required_columns.difference(data.columns)
if missing_columns:
    raise ValueError(f'Kolom tidak ditemukan: {sorted(missing_columns)}')

print('Ukuran dataset:', data.shape)
display(data.head())
print('Missing values:')
display(data[['YearsExperience', 'Salary']].isna().sum())
```

## 2. Menentukan fitur dan target

SVR memprediksi nilai kontinu. `YearsExperience` digunakan sebagai fitur dan `Salary` sebagai target.

```python
X = data[['YearsExperience']].to_numpy()
y = data['Salary'].to_numpy()

print('Bentuk X:', X.shape)
print('Bentuk y:', y.shape)
```

## 3. Membuat dan melatih model SVR

Model menggunakan kernel RBF dengan parameter `C=1000` dan `gamma=0.05` sebagaimana contoh latihan. Model kemudian dilatih menggunakan seluruh data.

```python
svr_model = SVR(C=1000, gamma=0.05, kernel='rbf')
svr_model.fit(X, y)
```

## 4. Visualisasi hasil prediksi

Data diurutkan berdasarkan lama pengalaman kerja agar garis prediksi tampil berurutan.

```python
order = np.argsort(X[:, 0])
X_sorted = X[order]
y_sorted = y[order]
y_pred_sorted = svr_model.predict(X_sorted)

plt.figure(figsize=(9, 5))
plt.scatter(X[:, 0], y, label='Data aktual')
plt.plot(X_sorted[:, 0], y_pred_sorted, label='Prediksi SVR')
plt.title('SVR: Prediksi Gaji berdasarkan Pengalaman Kerja')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 5. Evaluasi sederhana

MAE dan MSE mengukur besar kesalahan prediksi, sedangkan R² menunjukkan seberapa besar variasi target yang dijelaskan model. Karena evaluasi berikut dilakukan pada data yang sama dengan data training (*in-sample*), hasilnya bukan ukuran performa pada data baru.

```python
y_fit = svr_model.predict(X)
mae = mean_absolute_error(y, y_fit)
mse = mean_squared_error(y, y_fit)
r2 = r2_score(y, y_fit)

print(f'MAE: {mae:.3f}')
print(f'MSE: {mse:.3f}')
print(f'R² : {r2:.3f}')
```

## 6. Kesimpulan

SVR digunakan untuk memprediksi nilai kontinu, yaitu gaji berdasarkan pengalaman kerja. Grafik membandingkan data aktual dengan kurva prediksi SVR. Evaluasi pada data training saja tidak membuktikan model akan memiliki performa yang sama pada data baru.
